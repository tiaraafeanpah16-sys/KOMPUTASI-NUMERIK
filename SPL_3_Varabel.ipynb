```python
# MEMANGGIL LIBRARY DAN MEMBUAT MATRIKS
import numpy as np
# MEMASUKAN MATRIKS AUGMENTED
A = np.array([
    [3, 2, 4, -5],
    [-1, -1, 1, -6],
    [-3, 2, 6, -17]
], dtype=float)

A_awal = A.copy()

print("MATRIKS AUGMENTED AWAL:")
print(A)
# MENENTUKAN JUMLAH VARABEL DAN PIVOT
n = 3

for i in range(n):
# MENCARI PIVOT TERBESAR UNTUK MENGHINDARI PEMBAGIAN NOL
    max_row = i + np.argmax(np.abs(A[i:, i]))
# MENUKAR BARIS JIKA DI PERLUKAN
    A[[i, max_row]] = A[[max_row, i]]

    if abs(A[i, i]) < 1e-12:
        raise ValueError("Sistem tidak memiliki solusi tunggal.")

    pivot = A[i, i]
# MENGUBAH NILAI PIVOT MENJADI 1
    A[i] = A[i] / pivot
# MENGHILANGKAN ELEMEN LAIN PADA KOLOM PIVOT
    for j in range(n):
        if j != i:
            faktor = A[j, i]
            A[j] = A[j] - faktor * A[i]

    print(f"\nMatriks setelah langkah ke-{i+1}:")
    print(np.round(A, 4))
# MENGAMBIL HASIL PENYELESAIAN
x1 = A[0, 3]
x2 = A[1, 3]
x3 = A[2, 3]
# MENAMPILKAN HASIL
print("\nHASIL PENYELESAIAN SPL 3 VARIABEL")
print("--------------------------------")
print("Nilai x1 =", round(x1, 4))
print("Nilai x2 =", round(x2, 4))
print("Nilai x3 =", round(x3, 4))
#  PEMERIKSAAN KEBENARAN HASIL
hasil = A_awal[:, :3] @ np.array([x1, x2, x3])

print("\nPEMERIKSAAN HASIL:")
print("Hasil ruas kiri:", hasil)
print("Hasil ruas kanan:", A_awal[:, 3])

if np.allclose(hasil, A_awal[:, 3]):
    print("Kesimpulan: Hasil perhitungan BENAR.")
else:
    print("Kesimpulan: Hasil perhitungan TIDAK SESUAI.")
```
